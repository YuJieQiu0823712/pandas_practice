### Step 1. Import the necessary libraries:

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# print the graphs in the notebook
# % matplotlib inline

# set seaborn style to white
# sns.set_style('white')
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv). 

### Step 3. Assign it to a variable called tips

```python
tips = pd.read_csv("https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv",sep=',')
tips.head()
```

### Step 4. Delete the Unnamed 0 column

```python
del tips['Unnamed: 0']
```

### Step 5. Plot the total_bill column histogram

```python
ttbill = sns.histplot(tips.total_bill)
# distplot() or displot()

ttbill.set(xlabel='Value', ylabel='Frequency', title='Total Bill')

sns.despine() # take out the right and upper borders
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips) #
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.scatterplot(
    x='total_bill',
    y='tip',
    data=tips,
    size='size')
```

```python
sns.pairplot(tips) #
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.boxplot(y='day', x='total_bill', data=tips)
```

```python
sns.stripplot(x='day', y='total_bill', data=tips, jitter=True) # jitter prevent points from overlapping
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
sns.scatterplot(
    x='tip',
    y='day',
    hue='sex',
    data=tips,
)
```

```python
sns.swarmplot(
    x='tip',
    y='day',
    hue='sex',
    data=tips
)
```

```python
sns.stripplot(x='tip',y='day', hue='sex',data=tips, jitter=True)
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.boxplot(y='day', x='total_bill', hue='time', data=tips)
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
tips.head()
```

```python
sns.set(style='ticks')
g = sns.FacetGrid(tips, col='time')
g.map(sns.distplot, 'tip') # take series as inputs
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
g = sns.FacetGrid(tips,col='sex')
g.map_dataframe(sns.scatterplot, x='total_bill', y='tip', hue='smoker')
```

```python
g = sns.FacetGrid(tips,col='sex',hue='smoker')
g.map(plt.scatter, 'total_bill', 'tip', alpha=.7)
g.add_legend()
```
